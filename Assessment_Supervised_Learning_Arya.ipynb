### Limitations of Impurity-Based Feature Importance (MDI)


1. **Bias Towards High Cardinality Features**: Impurity-based feature importances are biased in favor of continuous numerical variables or categorical variables with high cardinality (many unique values). These variables provide more splitting opportunities in decision trees, making them artificially appear more "important" than they might actually be.
2. **Lack of Directionality**: Feature importances show *which* variables are influential, but they do not tell us the *direction* of the relationship (i.e., whether higher values of a feature increase or decrease the likelihood of subscription).
3. **Multicollinearity Sensitivity**: If two or more features are highly correlated, the tree algorithm can choose either of them to split the nodes. This dilutes the true importance of those features, spreading the significance score across them and underrepresenting their individual impact.
